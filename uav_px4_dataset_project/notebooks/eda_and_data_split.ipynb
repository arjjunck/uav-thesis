```python
from pathlib import Path
import pandas as pd

# Paste your actual CSV path here
DATA_PATH = Path("/Users/arjun/Documents/UAV thesis/uav_px4_dataset_project/processed_10hz/clean_ml_ready_10hz_expanded.csv")

print("Dataset path:")
print(DATA_PATH)

print("File exists:", DATA_PATH.exists())

if not DATA_PATH.exists():
    raise FileNotFoundError(f"File not found: {DATA_PATH}")

df = pd.read_csv(DATA_PATH, low_memory=False)

print("Loaded successfully")
print("Shape:", df.shape)
print("Unique flights:", df["log_id"].nunique())

print("\nDataset label distribution:")
print(df["dataset_label"].value_counts())

display(df.head())
```

```python
#Check flight-level balance and zero-current issue
current_col = "battery_status__current_a"

flight_summary = (
    df.groupby(["log_id", "dataset_label"])
    .agg(
        rows=("time_s", "count"),
        duration_s=("time_s", "max"),
        mean_current_a=(current_col, "mean"),
        max_current_a=(current_col, "max"),
        mean_power_w=("power_w", "mean"),
        max_power_w=("power_w", "max"),
        total_energy_wh=("cumulative_energy_wh", "max"),
        mean_target_60s=("target_energy_next_60s_wh", "mean")
    )
    .reset_index()
)

flight_summary["duration_min"] = flight_summary["duration_s"] / 60

# zero current and zero power percentage per flight
zero_stats = (
    df.assign(
        zero_current=df[current_col].abs() < 0.01,
        zero_power=df["power_w"].abs() < 0.01
    )
    .groupby("log_id")
    .agg(
        zero_current_percent=("zero_current", "mean"),
        zero_power_percent=("zero_power", "mean")
    )
    .reset_index()
)

zero_stats["zero_current_percent"] *= 100
zero_stats["zero_power_percent"] *= 100

flight_summary = flight_summary.merge(zero_stats, on="log_id", how="left")

print("Flight count by label:")
display(flight_summary["dataset_label"].value_counts())

print("Flight-level summary by label:")
display(
    flight_summary.groupby("dataset_label").agg(
        flights=("log_id", "count"),
        total_rows=("rows", "sum"),
        mean_duration_min=("duration_min", "mean"),
        median_duration_min=("duration_min", "median"),
        mean_power_w=("mean_power_w", "mean"),
        mean_total_energy_wh=("total_energy_wh", "mean"),
        mean_zero_current_percent=("zero_current_percent", "mean"),
        mean_zero_power_percent=("zero_power_percent", "mean")
    )
)

print("Flights with highest zero-current percentage:")
display(
    flight_summary.sort_values("zero_current_percent", ascending=False)
    .head(20)
)
```

```python
# Set practical thresholds
MAX_ZERO_CURRENT_PERCENT = 90
MIN_TOTAL_ENERGY_WH = 0.1
MIN_MAX_POWER_W = 5

usable_flights = flight_summary[
    (flight_summary["zero_current_percent"] < MAX_ZERO_CURRENT_PERCENT) &
    (flight_summary["total_energy_wh"] > MIN_TOTAL_ENERGY_WH) &
    (flight_summary["max_power_w"] > MIN_MAX_POWER_W)
].copy()

removed_flights = flight_summary[
    ~flight_summary["log_id"].isin(usable_flights["log_id"])
].copy()

print("Original flights:", flight_summary["log_id"].nunique())
print("Usable flights after filtering:", usable_flights["log_id"].nunique())
print("Removed flights:", removed_flights["log_id"].nunique())

print("\nUsable flights by label:")
display(usable_flights["dataset_label"].value_counts())

print("\nRemoved flights by label:")
display(removed_flights["dataset_label"].value_counts())

print("\nRows after filtering:")
df_clean = df[df["log_id"].isin(usable_flights["log_id"])].copy()

print("Original rows:", len(df))
print("Clean rows:", len(df_clean))
print("Removed rows:", len(df) - len(df_clean))

display(df_clean[[
    "log_id",
    "dataset_label",
    "time_s",
    "battery_status__voltage_v",
    "battery_status__current_a",
    "power_w",
    "cumulative_energy_wh",
    "target_energy_next_60s_wh"
]].head())
```
